### Ex 1.07 Data Plots

```python
# Large data set

import pandas as pd
import matplotlib.pyplot as plt

datasets = [
    ("sqrt-large.csv", "Square Root Error — Large Inputs"),
    ("sqrt-small.csv", "Square Root Error — Small Inputs")
]

for filename, title in datasets:
    df = pd.read_csv(filename)

    plt.figure(figsize=(10, 6))

    plt.plot(
        df["x"],
        df["original_error"],
        marker="o",
        label="Original sqrt"
    )

    plt.plot(
        df["x"],
        df["improved_error"],
        marker="x",
        label="Improved sqrt"
    )

    plt.xscale("log")
    plt.yscale("symlog", linthresh=0.000001)

    plt.xlabel("Input (x)")
    plt.ylabel("Percentage Error (%)")
    plt.title(title)

    plt.legend()
    plt.grid(True, which="both", alpha=0.3)
    plt.tight_layout()
    plt.show()
```
